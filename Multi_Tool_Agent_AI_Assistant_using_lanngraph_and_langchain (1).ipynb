```python
from google.colab import userdata
from google.colab import files

from langchain_google_genai import (
    ChatGoogleGenerativeAI,
    GoogleGenerativeAIEmbeddings,
)

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
# loading gemini

GOOGLE_API_KEY = userdata.get("GOOGLE_API_KEY")

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=GOOGLE_API_KEY,
    temperature=0
)
```

```python
# uploading file

from google.colab import files

uploaded = files.upload()
```

```python
loader = PyPDFLoader("/content/generative ai.pdf")

docs = loader.load()

print("Total Pages :", len(docs))
```

```python
# split documengts

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

splits = text_splitter.split_documents(docs)

print("Total Chunks :", len(splits))
```

```python
!pip install sentence-transformers
```

```python
# creating embedding

from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = FAISS.from_documents(
    documents=splits,
    embedding=embeddings
)

```

```python
# retriever
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 4}
)
```

```python
# testing retriever

query = "What is Generative AI?"

docs = retriever.invoke(query)

for i, doc in enumerate(docs):
    print(f"\n--------- Chunk {i+1} ---------\n")
    print(doc.page_content[:1000])
```

```python
# building rag chain

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template("""
You are a helpful AI assistant.

Answer the question only from the provided context.

Context:
{context}

Question:
{question}

Answer:
""")
```

```python
#creating chain

rag_chain = (
    {
        "context": lambda x: retriever.invoke(x),
        "question": lambda x: x
    }
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
#testing rag
```

```python
response = rag_chain.invoke(
    "Explain Generative AI in simple terms."
)

print(response)
```

```python
# lets convert into tool(agent 1)

from langchain.tools import tool

@tool
def rag_tool(query: str) -> str:
    """
    Use this tool for answering questions from uploaded PDFs.
    """
    return rag_chain.invoke(query)
```

```python
# test tool

rag_tool.invoke(
    "What are transformers?"
)
```

```python
from langchain.tools import tool
import sqlite3

@tool
def sql_tool(query: str):
    """Execute SQL query against SQLite database."""

    conn = sqlite3.connect("sales.db")
    cursor = conn.cursor()

    cursor.execute(query)
    result = cursor.fetchall()

    conn.close()

    return result
```

```python
# testing

sql_tool.invoke(
    "SELECT country, SUM(quantity) FROM sales GROUP BY country"
)
```

```python
from typing import TypedDict, Annotated
from langgraph.graph.message import add_messages
from langchain_core.messages import BaseMessage

class AgentState(TypedDict):
    messages: Annotated[list[BaseMessage], add_messages]
```

```python
from langgraph.graph import StateGraph, START, END
from langgraph.prebuilt import ToolNode
```

```python
# creating tool list

tools = [
    rag_tool,
    web_search_tool,
    python_tool,
    sql_tool
]
```

```python
# creating tool node
```

```python
tool_node = ToolNode(tools)
```

```python
# binding tools to llm

llm_with_tools = llm.bind_tools(tools)
```

```python
'''Now Gemini knows:

RAG tool exists
Web tool exists
Python tool exists
SQL tool exists'''
```

```python
# creating supervisor node
```

```python
def supervisor(state: AgentState):
    response = llm_with_tools.invoke(
        state["messages"]
    )

    return {
        "messages": [response]
    }
```

```python
# conditional routing
from langgraph.prebuilt import tools_condition
```

```python
# building graph

builder = StateGraph(AgentState)

builder.add_node(
    "supervisor",
    supervisor
)

builder.add_node(
    "tools",
    tool_node
)

builder.add_edge(
    START,
    "supervisor"
)

builder.add_conditional_edges(
    "supervisor",
    tools_condition,
    {
        "tools": "tools",
        END: END
    }
)

builder.add_edge(
    "tools",
    "supervisor"
)
```

```python
# compiling graph

graph = builder.compile()
```
